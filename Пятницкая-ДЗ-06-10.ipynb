```python
#Задача №1.
# Напишите программу, которая считывает с клавиатуры размер списка целых чисел. От 1 до размера списка сгенерировать случайным образом (random) массив.
# Вывести для сгенерированного массива:
# * длину
# * последний элемент
# * массив в обратном порядке (реверс)
# * если массив содержит числа 5 и 17 вывести “Да”, иначе “Нет”
# * создать копию массива и у копии удалить первый и последний элементы (использовать pop)

import random;

n = int(input())
a = []

for i in range(n):
    random_integer = random.randint(1, n);
    a.append(random_integer)

print("\n**Выход программы:**")
print("Введите размер массива:",a)
print("Длина массива:",len(a))
print("Последний элемент массива:",a[-1])
print("Массив в обратном порядке:",a[::-1])

if (5 or 17) in a:
    print("Да")
else:
    print("Нет")

b = a.copy()
b.pop(0)
b.pop(-1)
print("Копия массива после удаления первого и последнего элементов:",b)
```

```python
#Задача №2.
# Напишите программу, которая выполняет следующие действия:
# 1. Запрашивает у пользователя целочисленное число 𝑛
# 2. Вычисляет первые n элементов последовательности Фибоначчи. Последовательность Фибоначчи начинается с 0 и 1, и каждое последующее число в последовательности является суммой двух предыдущих чисел.
# 3. Находит сумму всех чётных чисел среди первых 𝑛 элементов последовательности Фибоначчи.
# 4. Выводит эту сумму на экран.

n = int(input())

a = [0, 1]

b = []

c = str()

s = 0

for i in range(n-2):
    a.append(a[i] + a[i+1])

for i in range(len(a)):
    if a[i]%2 == 0:
        b.append(a[i])
        s = s + a[i]
        c = c + str(a[i]) + ' + '

c = c[:-2] + "="

print("\nДля введённого пользователем числа n =", n)
print("Первые", n, "элементов последовательности Фибоначчи:",a)
print("Чётные числа среди этих элементов:",b)
print("Сумма этих чётных чисел:",c, s)
print("Программа должна вывести:",s)
```
